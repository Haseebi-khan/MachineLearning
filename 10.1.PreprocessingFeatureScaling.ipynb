# **Preprocessing Feature Scaling**

Feature Scaling

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(
    {
        'f1': [15, 18, 12, 10],
        'f2': [1, 3, 2, 5] 
    }
)
df
```

```python
df.describe()
```

## **MaxAbsScaler:**

```python
f1_maxabs = [x/abs(np.max(df.f1)) for x in df.f1]
f2_maxabs = [x/abs(np.max(df.f2)) for x in df.f2]

df1 = pd.DataFrame({'f1': f1_maxabs, 'f2': f2_maxabs})
df1
```

```python
df
```

```python
from sklearn.preprocessing import MaxAbsScaler
scaler = MaxAbsScaler()
arr = scaler.fit_transform(df)
df2 = pd.DataFrame(data=arr, columns=['f1','f2'])
df2
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

colors = ['red', 'green', 'blue', 'yellow']
ax.scatter(x=df.f1, y=df.f2, label="Before Scaling", color=colors, marker='*', s=100)
ax.scatter(x=df2.f1, y=df2.f2, label="After Scaling", color=colors, marker='o', s=60)
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.title("MaxAbsScaler")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```

## **MinMaxScaler**

```python
from sklearn.preprocessing import MinMaxScaler

scaler2 = MinMaxScaler()

arr2 = scaler2.fit_transform(df)

df3 = pd.DataFrame(data=arr2, columns=['f1', 'f2'])

df3
```

```python

fig, ax = plt.subplots()

colors = ['red', 'green', 'blue', 'yellow']
ax.scatter(x=df.f1, y=df.f2, label="Before Scaling", color=colors, marker='*', s=100)
ax.scatter(x=df3.f1, y=df3.f2, label="After Scaling", color=colors, marker='o', s=60)
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.title("MinMaxScaler")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```

## **Standard Scaler:**

```python
from sklearn.preprocessing import StandardScaler

scaler3 = StandardScaler()

arr3 = scaler3.fit_transform(df)

df4 = pd.DataFrame(data=arr3, columns=['f1', 'f2'])
df4
```

```python
fig, ax = plt.subplots()

colors = ['red', 'green', 'blue', 'yellow']
ax.scatter(x=df.f1, y=df.f2, label="Before Scaling", color=colors, marker='*', s=100)
ax.scatter(x=df4.f1, y=df4.f2, label="After Scaling", color=colors, marker='o', s=60)
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.title("Standard Scaler")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```

## **Robust Scaler**

```python
from sklearn.preprocessing import RobustScaler
scaler4 = RobustScaler()
arr4 = scaler4.fit_transform(df)
df5 = pd.DataFrame(data=arr4, columns=['f1', 'f2'])
df5
```

```python
fig, ax = plt.subplots()

colors = ['red', 'green', 'blue', 'yellow']
ax.scatter(x=df.f1, y=df.f2, label="Before Scaling", color=colors, marker='*', s=100)
ax.scatter(x=df5.f1, y=df5.f2, label="After Scaling", color=colors, marker='o', s=60)
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
plt.title("Robust Scaler")
plt.legend(loc="best")
plt.grid(True)
plt.show()
```

## **Effect Of Outlier Distribution:**

```python
import pandas as pd 
import numpy as np
import seaborn as sns
from sklearn import preprocessing
import matplotlib.pyplot as plt
```

```python
arr1 = np.concatenate([np.random.normal(20,2,1000), np.random.normal(1,2,25)])
arr2 = np.concatenate([np.random.normal(40,2,1000), np.random.normal(70,2,25)])

df = pd.DataFrame({'f1': arr1, 'f2': arr2})
df
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.set_title("Before Scaling")
sns.kdeplot(df["f1"], color='r')
sns.kdeplot(df["f2"], color='b')
plt.show
```

```python
scaler = preprocessing.MinMaxScaler()
minmax_nparr = scaler.fit_transform(df)
minmax_df = pd.DataFrame(minmax_nparr, columns=['f1', 'f2'])\
    
scaler2 = preprocessing.StandardScaler()
Standard_nparr = scaler2.fit_transform(df)
standard_df = pd.DataFrame(Standard_nparr, columns=['f1', 'f2'])


scaler3 = preprocessing.RobustScaler()
robust_nparr = scaler3.fit_transform(df)
robust_df = pd.DataFrame(robust_nparr, columns=['f1', 'f2'])
```

```python
fig, (ax1, ax2, ax3, ax4) = plt.subplots(ncols=4, figsize=(20,5))

ax1.set_title("Before Scaling")
sns.kdeplot(df["f1"], color='r', ax=ax1)
sns.kdeplot(df["f2"], color='b', ax=ax1)


ax1.set_title("After Min Max Scaling")
sns.kdeplot(minmax_df["f1"], color='black', ax=ax2)
sns.kdeplot(minmax_df["f2"], color='g', ax=ax2)


ax1.set_title("After Standard Scaling")
sns.kdeplot(standard_df["f1"], color='black', ax=ax3)
sns.kdeplot(standard_df["f2"], color='g', ax=ax3)


ax1.set_title("After Robust Scaling")
sns.kdeplot(robust_df["f1"], color='r', ax=ax4)
sns.kdeplot(robust_df["f2"], color='b', ax=ax4)

plt.show()
```

## **Applying Concepts on Real Datasets:**

```python
df = pd.read_csv(r'D:\Codes\MachineLearning\Datasets\scaling-data1.csv')
df
```

```python
df.describe()
```

```python
df.info()
```

```python
x = df.drop('target', axis=1)
x
```

```python
y = df["target"]
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=54)
len(x_train), len(x_test), len(y_train), len(y_test)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python

from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

ss.fit(x_train[['age', 'salary']])

x_train_scaler = ss.transform(x_train[['age', 'salary']])
x_test_scaler = ss.transform(x_test[['age', 'salary']])

x_train_scaler.shape, x_train_scaler.shape 
```

```python
x_train_scalered = pd.DataFrame(x_train_scaler, columns=x_train.columns)
x_test_scalered = pd.DataFrame(x_test_scaler, columns=['age','salary'])

x_train_scalered
```

```python
x_test_scalered
```

```python
np.round(x_train_scalered.describe(),2)
```

```python
np.round(x_test_scalered.describe(),2)
```

```python
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2, ncols=2, figsize=(25,15))

ax1.set_title("Age Before Scaling")
sns.kdeplot(x_train['age'], ax=ax1, color='red')

ax2.set_title("Age After Scaling")
sns.kdeplot(x_train_scalered['age'], ax=ax2, color='red')

ax3.set_title("Salary Before Scaling")
sns.kdeplot(x_train['salary'], ax=ax3, color='blue')

ax4.set_title("Salary After Scaling")
sns.kdeplot(x_train_scalered['salary'], ax=ax4, color='blue')

plt.show()
```
