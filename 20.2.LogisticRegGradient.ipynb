```python
import numpy as np
import matplotlib.pyplot as plt

# Define the sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Define its derivative
def sigmoid_derivative(z):
    s = sigmoid(z)
    return s * (1 - s)

# Create a range of z values
z = np.linspace(-10, 10, 400)

# Compute sigmoid and derivative
s = sigmoid(z)
ds = sigmoid_derivative(z)

# Plot both
plt.figure(figsize=(8, 5))
plt.plot(z, s, label='Sigmoid Function', linewidth=2)
plt.plot(z, ds, label='Sigmoid Derivative', linestyle='--', linewidth=2)
plt.title('Sigmoid Function and its Derivative')
plt.xlabel('z')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def sigmoid(z):
    return 1 / ( 1 + np.exp(-z) )
```

```python
import numpy as np
def gradien_descent(X, y):
    # Add bias term (column of 1s at index 0)
    X = np.insert(X, 0, 1, axis=1)
    
    # Initialize weights (including bias weight)
    weights = np.ones(X.shape[1])
    
    lr = 0.1  # Learning rate
    epocs = 1000  # Number of iterations
    
    for i in range(epocs):
        
        # Compute prediction
        y_hat = sigmoid(np.sum(np.dot(X, weights)))
        
        # Update weights
        weights = weights - lr * (1/X.shape[0]) * np.dot((y_hat - y) ,X)
    
    # Return bias weight and feature weights separately
    return weights[0], weights[1:]
```

```python
bais, weights = gradien_descent(X, y)
print("w0: ", bais)
print("w1 and w2: ", weights)
```

```python
m = -(weights[0]/weights[1])
b = -(bais/weights[1])

# Generate x values from -3 to 3
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b

plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100)
plt.ylim(-3, 2)

# Show the plot
plt.show()
```
