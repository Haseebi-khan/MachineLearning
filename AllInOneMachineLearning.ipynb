Machine Learning Using Graient Descent Algorithm

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import preprocessing
from numpy import set_printoptions

df = pd.read_csv("study-hours.csv")
df
```

```python
scaler = preprocessing.MinMaxScaler(feature_range=(0, 4))

rescaled = scaler.fit_transform(df[["Scores"]])
# why this line is not working.
# set_printoptions(precision=2)

df["GPA"] = rescaled
df
```

```python
X = df[['Hours']]
y= df['GPA']

plt.scatter(X, y, color='black', label='GPA')
plt.title('Graient Descent Algorithm')
plt.xlabel('Hours')
plt.ylabel('GPA')
plt.grid()
plt.legend()
plt.show()
```

```python
df.columns
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X,y)
```

```python
print(lr.intercept_)
print(lr.coef_[0])
```

```python
y_pred = lr.predict(df[['Hours']])
print("Test Value: ",df['Hours'])
print("Predicted Value: ", y_pred)
```

```python
plt.scatter(X, y, color='black',label='GPA')
plt.plot(df['Hours'], y_pred, color='red', linewidth=2, label='Best Line Fit')
plt.xlabel('Hours')
plt.ylabel('GPA')
plt.title("Linear Regression Model.")
plt.grid(True)
plt.legend()
plt.show()
```

```python

a = lr.intercept_
b = lr.coef_[0]

ybar = np.mean(y)
yhat = a + b * X['Hours']  # x["study_hours"] is a Series

mae = (np.sum(np.abs(y - yhat))) / len(y) #
mse = (np.sum((y - yhat)**2)) / len(y)
rmse = np.sqrt((np.sum((y - yhat)**2)) / len(y))
sse = np.sum((y - yhat)**2)
sst = np.sum((y - ybar)**2)
r2 = 1 - sse/sst

# ==================================================================
# MAE: Mean Absolute Errors
# MSE: Mean Squared Error
# RMSE: Root Mean Squared Error
# SSE: S um of Squared Errors
# SST: Total Sum of Squares
# R2: R-squared (also known as the coefficient of determination)
# ==================================================================

print(f"MAE: ", mae)
print(f"MSE: ", mse)
print(f"RMSE: ", rmse)
print(f"R2: ", r2)
```

```python
# Trim outliers
q_low = data["Income"].quantile(0.05)
q_high = data["Income"].quantile(0.95)

trimmed_data = data[(data["Income"] >= q_low) & (data["Income"] <= q_high)]

print("\nAfter Trimming (Outliers Removed):\n", trimmed_data)
```

🔹 2. Capping / Winsorization (Replace Outliers)

```python
from scipy.stats.mstats import winsorize

# Convert to numpy array
income = data["Income"].values

# Apply winsorization: 5% from both sides
capped_income = winsorize(income, limits=[0.05, 0.05])

data["Capped_Income"] = capped_income

print("\nAfter Capping (Outliers Replaced):\n", data)
```

🔹 3. Discretization (Binning)

```python
bins = [0, 40, 70, 500]  # Define range bins
labels = ["Low", "Medium", "High"]

data["Income_Level"] = pd.cut(data["Income"], bins=bins, labels=labels)

print("\nAfter Discretization (Binning):\n", data)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder , OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer


df = pd.read_csv("salary-package1.csv")
df
```

```python
oe = OrdinalEncoder(categories=[['Low', 'Medium', 'High']], dtype=np.int8)
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)
si = SimpleImputer(strategy='mean', missing_values=np.nan)
X = df.drop('salary', axis=1)
y = df['salary']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
transformer1 = ColumnTransformer(
    transformers=[
        ('tmf1', ohe,['gender']),
        ('tmf2', si, ['cgpa']),
        ('tmf3', oe, ['iq'])
    ],
    remainder='passthrough'
)
```

```python
transformer1.fit(X_train)
```

```python
arr_X_train =  transformer1.transform(X_train)
arr_X_test =  transformer1.transform(X_test)
arr_X_test
```

```python
from sklearn.linear_model import LinearRegression
lr2 = LinearRegression()
lr2.fit(arr_X_train, y_train)
```

```python
test_features =  np.array(['Male', 3.5, 'High']).reshape(1,3)
test_features_transform =  np.array([1, 3.5, 2]).reshape(1,3)
test_features, test_features_transform
```
