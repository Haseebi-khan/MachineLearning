# **CHAPTER 6: Decision Trees**

Training and Visualizing a Decision Tree

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

iris = load_iris()
X = iris.data[:, 2:]
y = iris.target

tree_clf = DecisionTreeClassifier(max_depth=2)
tree_clf.fit(X, y)
```

```python
from sklearn.tree import export_graphviz

export_graphviz(
    tree_clf,
    out_file="iris_tree.dot",
    feature_names=iris.feature_names[2:],
    class_names=iris.target_names,
    rounded=True,
    filled=True
)
```

```python
!dot -Tpng iris_tree.dot -o iris_tree.png
```

Estimating Class Probabilities

```python
tree_clf.predict_proba([[9,5]])
```

```python
tree_clf.predict([[5, 1.5]])
```

# **Regression**

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(max_depth=2)
tree_reg.fit(X, y)
```

```python
!dot -Tpng iris_tree_reg.dot -o iris_tree_reg.png
```

```python
tree_reg.predict([[5,1.8]])
```

**2. Is a node’s Gini impurity generally lower or greater than its parent’s? Is it gener‐
ally lower/greater, or always lower/greater?**

Ans: A node's Gini impurity is generally lower than its parent's because Decision Trees select splits that reduce impurity. However, an individual child node is not always less impure than its parent. What is guaranteed is that the weighted average Gini impurity of the child nodes is less than or equal to the impurity of the parent node.
