```python
import pandas as pd

# Load dataset
icd_df = pd.read_csv(
    "/home/noneo/Codes/ML/MachineLearning/Hand_On_ML/Data/Healthcare Cost Dataset/train/train/icd_df_train.csv"
)

# Clean diagnosis codes
icd_df["Diagnosis_Code"] = (
    icd_df["Diagnosis_Code"]
    .astype(str)
    .str.upper()
    .str.strip()
)

# Extract the first 3 characters (ICD-10 category)
icd_df["ICD_Prefix"] = icd_df["Diagnosis_Code"].str[:3]

# Count frequency of each prefix
prefix_counts = (
    icd_df["ICD_Prefix"]
    .value_counts()
    .reset_index()
)

prefix_counts.columns = ["ICD_Prefix", "Count"]



print(f"\nTotal unique prefixes: {len(prefix_counts)}")
```
